```python
import pandas as pd

df_raw = pd.read_csv('../data/raw/telco_churn.csv')
df_clean = pd.read_csv('../data/processed/telco_churn_clean.csv')

print("Raw shape:", df_raw.shape)
print("Clean shape:", df_clean.shape)
print("\nRaw TotalCharges dtype:", df_raw['TotalCharges'].dtype)
print("Clean totalcharges dtype:", df_clean['totalcharges'].dtype)
```

```python
print("Missing values in clean data:")
print(df_clean.isnull().sum())
```

```python
print("High risk customers:", df_clean['is_high_risk'].sum())
print("Churn distribution:")
print(df_clean['churn'].value_counts())
```

```python
churn_by_risk = df_clean.groupby('is_high_risk')['churn'].mean() * 100
print("Churn rate by risk level:")
print(churn_by_risk.round(2))
```
