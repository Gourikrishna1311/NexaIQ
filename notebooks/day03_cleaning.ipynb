```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/telco_churn.csv')
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
```

```python
problem_rows = df[df['TotalCharges'].str.strip() == '']
print("Rows with empty TotalCharges:", len(problem_rows))
print(problem_rows[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
categorical_cols = ['gender', 'Partner', 'Dependents', 'PhoneService',
                    'InternetService', 'Contract', 'PaymentMethod', 'Churn']
for col in categorical_cols:
    print(f"{col}: {df[col].unique()}")
```
