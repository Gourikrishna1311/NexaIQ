```python
import numpy as np
import pandas as pd
import os

os.chdir('C:/Users/gouri/Desktop/NexaIQ')

df = pd.read_csv('data/processed/telco_churn_clean.csv')

charges = np.array(df['monthlycharges'])
tenure = np.array(df['tenure'])
total_value = np.array(df['total_value'])

print("Monthly Charges Statistics:")
print(f"  Mean:      ${np.mean(charges):.2f}")
print(f"  Median:    ${np.median(charges):.2f}")
print(f"  Std Dev:   ${np.std(charges):.2f}")
print(f"  Min:       ${np.min(charges):.2f}")
print(f"  Max:       ${np.max(charges):.2f}")
print(f"  25th pct:  ${np.percentile(charges, 25):.2f}")
print(f"  75th pct:  ${np.percentile(charges, 75):.2f}")
```

```python
churned = df[df['churn'] == 1]
retained = df[df['churn'] == 0]

churned_charges = np.array(churned['monthlycharges'])
retained_charges = np.array(retained['monthlycharges'])

print("Churned customers:")
print(f"  Avg monthly charge: ${np.mean(churned_charges):.2f}")
print(f"  Avg tenure: {np.mean(np.array(churned['tenure'])):.1f} months")
print(f"  Avg total value: ${np.mean(np.array(churned['total_value'])):.2f}")

print("\nRetained customers:")
print(f"  Avg monthly charge: ${np.mean(retained_charges):.2f}")
print(f"  Avg tenure: {np.mean(np.array(retained['tenure'])):.1f} months")
print(f"  Avg total value: ${np.mean(np.array(retained['total_value'])):.2f}")
```

```python
features = ['tenure', 'monthlycharges', 'totalcharges',
            'total_value', 'is_high_risk', 'seniorcitizen']

corr_matrix = np.corrcoef([df[col].values for col in features])

print("Correlation with churn:")
for i, col in enumerate(features):
    churn_corr = np.corrcoef(df[col].values, df['churn'].values)[0, 1]
    print(f"  {col}: {churn_corr:.4f}")
```

```python
total_monthly_revenue = np.sum(df['monthlycharges'].values)
high_risk_revenue = np.sum(df[df['is_high_risk'] == 1]['monthlycharges'].values)
churned_revenue_lost = np.sum(df[df['churn'] == 1]['monthlycharges'].values)

print(f"Total monthly revenue:        ${total_monthly_revenue:,.2f}")
print(f"Revenue from high risk:       ${high_risk_revenue:,.2f}")
print(f"Monthly revenue already lost: ${churned_revenue_lost:,.2f}")
print(f"% revenue at risk:            {(high_risk_revenue/total_monthly_revenue)*100:.2f}%")
print(f"% revenue already lost:       {(churned_revenue_lost/total_monthly_revenue)*100:.2f}%")
```

```python
charge_bins = np.linspace(df['monthlycharges'].min(),
                          df['monthlycharges'].max(), 10)

hist, bin_edges = np.histogram(df['monthlycharges'], bins=charge_bins)

print("Monthly charge distribution:")
for i in range(len(hist)):
    print(f"  ${bin_edges[i]:.0f} - ${bin_edges[i+1]:.0f}: {hist[i]} customers")
```
